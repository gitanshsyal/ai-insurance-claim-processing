```python
from google.colab import userdata
from groq import Groq
from collections import Counter

client = Groq(api_key=userdata.get("groq_api_key"))
MODEL = "openai/gpt-oss-120b"

def ask_many_times(prompt, runs=15):
    answers = []
    for _ in range(runs):
        response = client.chat.completions.create(
            model=MODEL,
            messages=[{"role": "user", "content": prompt}],
            temperature=1.0,          # same temperature every time
            max_tokens=2000,          # plenty of room for thinking + answer
            reasoning_effort="low",   # think briefly
            include_reasoning=False,  # return only the final answer
        )
        answer = (response.choices[0].message.content or "").strip().lower().rstrip(".!")
        answers.append(answer or "(empty)")
    return Counter(answers)

def show(title, counts):
    print(f"\n{title}  ->  {len(counts)} different answers")
    for answer, count in counts.most_common():
        print(f"  {'█' * count} {count}  {answer}")
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
handbook=open('policy.txt').read()
chunks_splitter=RecursiveCharacterTextSplitter(chunk_size=500,chunk_overlap=50)
chunks=chunks_splitter.create_documents([handbook])
print(len(chunks))
for chunk in chunks[:10]:
    print(chunk.page_content, "\n-----")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
```

```python
from langchain_core.vectorstores import InMemoryVectorStore

# InMemoryVectorStore, a lightweight vector store that keeps all embeddings and documents in the system memory (RAM).

vector_store = InMemoryVectorStore.from_documents(chunks, embeddings)
```

```python
from typing import Literal
from pydantic import BaseModel, ConfigDict
class policy_detail(BaseModel):
  model_config = ConfigDict(extra="forbid")
  is_claim_valid:bool
  coverage_limit:float
  deductible:float
  eligible_loss:float
  claim_amount:float
```

```python
from typing import Literal
from pydantic import BaseModel, ConfigDict
class claim_valid(BaseModel):
  model_config = ConfigDict(extra="forbid")
  claim_valid:bool
  reason:str
```

```python
def check_policy(user_data):
  query = user_data.model_dump_json()
  results = vector_store.similarity_search_with_score(query, k=3)
  context = "\n\n".join(
        doc.page_content for doc, score in results
    )
  system=f'''you are working in insurance company and you will receive user details. you have handbook that contains entire details of policies that are available
  and you need to ans by seeing data based on it.You need to select the relevant policy based on the insurance type.you need to Check whether the claim is covered under the policy.
  you need to  if claim is eligible or not and also provide the valid reason.Do not invent new data and provide
  false information.
  <context>
  {context}</context>'''
  response=client.chat.completions.create(
  model=MODEL,
  messages=[{'role':'system','content':system},{'role':'user','content':query}],
  reasoning_effort='low',
  response_format={                     # 👈 the only new part
            "type": "json_schema",
            "json_schema": {"name": "ticket1", "schema":claim_valid.model_json_schema()},
        },
  )
  result =claim_valid.model_validate_json(response.choices[0].message.content)
  return result
```

```python
from typing import Literal
from pydantic import BaseModel, ConfigDict
class claim(BaseModel):
   model_config = ConfigDict(extra="forbid")
   customer_name:str
   policy_number:str
   event_date:str
   event_location:str
   event_description:str
   loss_damage_details:str
   claimed_amount:float| None
   supporting_documents:list[str]
   customer_request:str
```

```python
class insurance_type(BaseModel):
    insurance_type:Literal[
        "Home Insurance",
        "Vehicle Insurance",
        "Medical Insurance",
        "Life Insurance"
    ]
    reason:str
```

```python
class claim_metrics(BaseModel):
  model_config = ConfigDict(extra="forbid")
  eligible_loss:float
  coverage_limt:float
  deductible:float
  claim_amount:float
```

```python
def extract_claim_metrics(result):
  query =result.model_dump_json()
  results = vector_store.similarity_search_with_score(query, k=3)
  context = "\n\n".join(
        doc.page_content for doc, score in results
  )
  system=f'''you are working in insurance company and you will receive user details. you have handbook that contains entire details of policies that are available
  and you need to ans by seeing data based on it. you need to extract Eligible Loss, Coverage Limit and Deductible from the handbook upon the matching policy.
  Do not invent new data and do not provide false information.claim_amount from the user input.
  <context>{context}</context>
  '''
  response=client.chat.completions.create(
      model=MODEL,
      messages=[{'role':'system','content':system},{'role':'user','content':query}],
      reasoning_effort='low',
      response_format={                     # 👈 the only new part
            "type": "json_schema",
            "json_schema": {"name": "ticket1", "schema":claim_metrics.model_json_schema()},
        },
  )
  result =claim_metrics.model_validate_json(response.choices[0].message.content)
  return result
```

```python
def calculate_claim_amount(data):
    potential_payout =min(data.eligible_loss, data.coverage_limt)-data.deductible
    return potential_payout
```

```python

def assement_email(output, result1, payable_amount, claim_valid):

    print("Claim details:", result1)
    print("Claim validity:", claim_valid)

    if claim_valid is True:

        system = """
        You are an insurance company assistant.

        Generate a personalized customer email in fewer than 60 words.

        Include:
        - Claim summary
        - Insurance type
        - Policy number
        - Claimed amount
        - Coverage limit
        - Deductible
        - Approved payable amount

        Assessment Outcome:
        Approved if the claim satisfies policy conditions.
        Pending if required information is missing or unclear.

        After the email, provide the final outcome and
        explain the reason in 2-3 lines.
        """

        user = f"""
        Customer details: {output}
        Claim details: {result1}
        Net payable amount: {payable_amount}
        """

    else:

        system = """
        You are an insurance company assistant.

        Generate a personalized and polite rejection email
        in fewer than 60 words.

        Explain that the claim could not be approved
        based on the policy assessment.

        Include:
        - Assessment Status: Rejected
        - Reason for rejection

        Do not include payout calculations or claim
        payment details.

        After the email, provide the final outcome
        with the reason in 2-3 lines.
        """

        user = f"""
        Customer details: {output}

        Claim validity: {claim_valid}

        Generate a rejection email based on the
        available claim information.
        """

    # API call is shared by both conditions
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": system},
            {"role": "user", "content": user}
        ],
        reasoning_effort="low"
    )

    return response.choices[0].message.content
```

```python
def valid_data(output):
    required_fields = {
        "customer_name": output.customer_name,
        "policy_number": output.policy_number,
        "event_date": output.event_date,
        "event_description": output.event_description,
        "loss_damage_details": output.loss_damage_details,
        "claim_amount": output.claimed_amount
    }

    missing_fields = [
        field
        for field, value in required_fields.items()
        if value is None
        or (isinstance(value, str) and value.strip().upper() in ["", "N/A", "NA", "NONE"])
    ]

    if not missing_fields:

        print("All required claim details are available. \nThe claim proceeds to the next stage. ")
        print(' '*30)
        print(check_insurance_type(output))
        print(''*35)
        result=check_policy(output)
        if result.claim_valid:
           result1=extract_claim_metrics(output)
           print(result1)
           print(''*40)
           payable_amount=calculate_claim_amount(result1)
           print('Estimated Payable Amount: '+str(payable_amount))
           print(''*40)
           print('Email\n'+'===='*25)
           assessment=assement_email(output,result1,payable_amount,result.claim_valid)
           print(assessment)
        else:
             # Rejected claim: no need to calculate payout metrics
            assessment = assement_email(
            output,
            None,
            None,
            result.claim_valid
            )
            print(''*40)
            print('Email\n'+'===='*25)
            print(assessment)
    else:
        print("Missing fields:", missing_fields)
        print(' '*30)
        print(send_email(output,missing_fields))
```
